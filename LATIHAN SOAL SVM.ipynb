```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC

# 202431181_Zhalsabila Amalia Putri
```

```python
X = np.array([
    [3, 3],
    [4, 4],
    [1, 1],
    [2, 1]
])


y = np.array([1, 1, -1, -1])


X_baru = np.array([[4, 2]])

print("Data latih:")
print(X)

print("\nLabel kelas:")
print(y)

print("\nData uji:")
print(X_baru)

# 202431181_Zhalsabila Amalia Putri
```

```python
# 202431181_Zhalsabila Amalia Putri

model = SVC(kernel='linear', C=1000000)


model.fit(X, y)


print("Bobot w:", model.coef_[0])
print("Bias b:", model.intercept_[0])


print("\nSupport Vector:")
print(model.support_vectors_)
```

```python
w1 = model.coef_[0][0]
w2 = model.coef_[0][1]
b = model.intercept_[0]

print("Persamaan hyperplane:")
print(f"{w1:.2f}x1 + {w2:.2f}x2 + ({b:.2f}) = 0")

print("\nPersamaan garis:")
print(f"x2 = {-b/w2:.2f} - ({w1/w2:.2f})x1")

# 202431181_Zhalsabila Amalia Putri
```

```python
# Membuat bidang grafik
x1 = np.linspace(0, 6, 100)
x2 = -(w1 * x1 + b) / w2


kelas_positif = X[y == 1]
kelas_negatif = X[y == -1]

plt.scatter(
    kelas_positif[:, 0],
    kelas_positif[:, 1],
    color='blue',
    marker='o',
    label='Kelas +1'
)

plt.scatter(
    kelas_negatif[:, 0],
    kelas_negatif[:, 1],
    color='red',
    marker='x',
    label='Kelas -1'
)

# Menampilkan support vector
plt.scatter(
    model.support_vectors_[:, 0],
    model.support_vectors_[:, 1],
    s=180,
    facecolors='none',
    edgecolors='green',
    label='Support Vector'
)


plt.plot(x1, x2, color='black', label='Hyperplane')

plt.scatter(
    X_baru[:, 0],
    X_baru[:, 1],
    color='orange',
    marker='*',
    s=200,
    label='Data Uji'
)

plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Visualisasi Linear SVM')
plt.grid(True)
plt.legend()
plt.show()

# 202431181_Zhalsabila Amalia Putri
```

```python
hasil_prediksi = model.predict(X_baru)

nilai_fungsi = model.decision_function(X_baru)

print("Data uji:", X_baru[0])
print("Nilai fungsi keputusan:", nilai_fungsi[0])
print("Hasil klasifikasi:", hasil_prediksi[0])

if hasil_prediksi[0] == 1:
    print("Data termasuk kelas +1")
else:
    print("Data termasuk kelas -1")

# 202431181_Zhalsabila Amalia Putri
```
